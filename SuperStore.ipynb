### Handle outliers in numerical columns (Sales, Profit, Quantity) using IQR method


```python
target_columns=['Sales','Profit','Quantity']
outlier_summary=[]

for col in target_columns:
    Q1=df[col].quantile(0.25)
    Q3=df[col].quantile(0.75)
    IQR=Q3-Q1

    lower_bound=Q1-1.5*IQR
    upper_bound=Q3+1.5*IQR

    outliers=(df[col]<lower_bound) | (df[col]>upper_bound)
    outlier_count=outliers.sum()
    
    outlier_summary.append({
        'Column': col,
        'Q1 (25%)': round(Q1, 2),
        'Q3 (75%)': round(Q3, 2),
        'IQR': round(IQR, 2),
        'Lower Bound': round(lower_bound, 2),
        'Upper Bound': round(upper_bound, 2),
        'Count':outlier_count
    })
outlier_df = pd.DataFrame(outlier_summary)
display(outlier_df)
        
        
```

```python
# Handling the Outlier
for col in target_columns:
    col_median=df[col].median()
    df[col]=np.where(outliers,col_median,df[col])
```

### Distribution analysis: skewness and kurtosis for key numerical features


```python
skewness=df[numeric_features].skew()
kutosis=df[numeric_features].kurtosis()
summary_skewness=pd.DataFrame({
    'Skewness':skewness,
    'Kurtosis':kutosis
})
print('--Skewness and Kurtosis Summaryy --')
display(summary_skewness.round(3))
    
```

```python
print("\n--DISTRIBUTION INTERPRETATION --")
for col in numeric_features:
    s = summary_skewness.loc[col, 'Skewness']
    k = summary_skewness.loc[col, 'Kurtosis']
    
    skew_desc = "Right-skewed (Positive)" if s > 0.5 else ("Left-skewed (Negative)" if s < -0.5 else "Symmetrical")
    kurt_desc = "Leptokurtic (Heavy-tailed)" if k > 0 else ("Platykurtic (Light-tailed)" if k < 0 else "Mesokurtic (Normal)")
    
    print(f"• {col}: Skewness = {s:.2f} ({skew_desc}) | Kurtosis = {k:.2f} ({kurt_desc})")
```

### Find which regions are underperforming

```python
regional_perf=df.groupby('Region').agg(
    Total_sales=('Sales','sum'),
    Total_profit=('Profit','sum'),
    Order_count=('Row ID','count'),
    Avg_discount=('Discount','mean')
   ).reset_index()
regional_perf['Profit_Margin (%)']=(regional_perf['Total_profit']/regional_perf['Total_sales'])*100
regional_perf.sort_values(by='Total_profit',ascending=True)
print("---REGIONAL PERFORMANCE RANKING (LOWEST TO HIGHEST PROFIT) ---")
display(regional_perf.round(2))
lowest_region = regional_perf.iloc[0]
print(f"\nUnderperforming Region: {lowest_region['Region']}")
print(f"• Total Profit: ${lowest_region['Total_profit']:,.2f}")
print(f"• Profit Margin: {lowest_region['Profit_Margin (%)']:.2f}%")
print(f"• Average Discount: {lowest_region['Avg_discount']*100:.2f}%")
```

### Analyze the relationship between discount and profit

```python
discount_profit_correlation=df['Discount'].corr(df['Profit'])
print(f"Correaltion between Discount and Profit: {discount_profit_correlation:.4f}")
```

```python
discount_analysis = (
    df.groupby('Discount')
    .agg(
        Order_Count=('Sales', 'count'),
        Total_Sales=('Sales', 'sum'),
        Total_Profit=('Profit', 'sum'),
        Avg_Profit=('Profit', 'mean'),
        Avg_Sales=('Sales', 'mean'),
    )
    .reset_index()
)
discount_analysis['Profit_Margin (%)'] = (
    discount_analysis['Total_Profit'] / discount_analysis['Total_Sales']
) * 100
print('--- PROFITABILITY BY DISCOUNT TIER ---')
display(discount_analysis.round(2))

```

```python
plt.figure(figsize=(10,6))
sns.barplot(
        discount_analysis,
        x='Discount',
        y='Avg_Profit',
        
)
plt.title('Average Profit by Discount Level', fontsize=14, fontweight='bold')
plt.xlabel('Discount Rate', fontsize=12)
plt.ylabel('Average Profit ($)', fontsize=12)
plt.axhline(0, color='black', linestyle='--', linewidth=0.8)

plt.tight_layout()
plt.savefig('Correlation  between Discount and Sales .png', dpi=300, bbox_inches='tight')
plt.show()
```

### Provide 5 actionable business recommendations

1.Cap Discount Tiers at 15% Max
--------------------------------------------------------------------------------------------------------------------------------------------
Finding: Orders that offer discounts of twenty percent or more result in average profits because the margin is    eroded heavily (r = negative zero point thirty eight).

Action: Put automated controls into the checkout process so that the discount for orders never goes above fifteen percent. For any enterprise custom quote that goes beyond fifteen percent the regional director must approve it.

2.Capitalize on Q4 Demand Spikes
--------------------------------------------------------------------------------------------------------------------------------------------
Finding: Sales surge dramatically during Q4 (September, November and December) while Q1 (January/February) experiences revenue drops.

Action: Align inventory procurement and warehouse staffing to peak in Q3. Launch targeted campaigns in early Q1 to smooth out seasonal revenue dips and liquidate slow‑moving inventory.

3.Restructure Strategy in Underperforming Regions
--------------------------------------------------------------------------------------------------------------------------------------------
Finding: Central and Southern regions underperform in profit margins despite gross sales largely due to high baseline discounting.

Action: Re‑evaluate freight pricing models and audit local discounting practices. Shift sales incentives in low‑margin regions from gross volume target metrics to margin targets.

4.Expand High‑Margin Sub‑Categories
--------------------------------------------------------------------------------------------------------------------------------------------
Finding: Technology (Copiers, Phones) and Office Supplies (Paper, Envelopes) generate the overall profit margins whereas Furniture (Tables, Bookcases) operates near net loss.

Action: Reallocate marketing budget toward top‑performing Technology and Office Supplies categories. Renegotiate supplier. Increase retail price points on low‑margin Furniture lines.

5.Cross‑Sell to Corporate & Consumer Segments
--------------------------------------------------------------------------------------------------------------------------------------------
Finding: Consumer and Corporate segments account for over eighty percent of order volume and display high average basket sizes.

Action: Implement automated cross‑selling algorithms at checkout (for example bundling high‑margin Office Supplies like Paper, with Technology hardware purchases) to increase order value without offering deep discounts.
